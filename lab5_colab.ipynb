```python
def bubble_sort(arr):
    if not isinstance(arr, list):
        raise TypeError("Потрібно список")

    swaps = 0
    n = len(arr)

    for i in range(n):
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swaps += 1

    return arr, swaps
```

```python
print(bubble_sort([5,2,9,1]))
print(bubble_sort([3,1,2]))
print(bubble_sort([10,8,6]))
```

```python
try:
    print(bubble_sort("abc"))
except Exception as e:
    print("Помилка:", e)
```

```python
import matplotlib.pyplot as plt

arrays = [[5,2,9,1],[3,1,2],[10,8,6],[1,2,3]]
swaps = []

for arr in arrays:
    _, s = bubble_sort(arr.copy())
    swaps.append(s)

plt.bar(["1","2","3","4"], swaps)
plt.title("Bubble Sort")
plt.show()
```
